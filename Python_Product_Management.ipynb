```python
products = {}

def load_dummy_data():
    products[1] = {"name": "Laptop", "price": 55000}
    products[2] = {"name": "Mouse", "price": 500}
    products[3] = {"name": "Keyboard", "price": 1500}
    products[4] = {"name": "Monitor", "price": 12000}
    products[5] = {"name": "Mobile", "price": 25000}
    products[6] = {"name": "Tablet", "price": 20000}
    products[7] = {"name": "Printer", "price": 8000}
    products[8] = {"name": "Speaker", "price": 3000}
    products[9] = {"name": "Headphone", "price": 2000}
    products[10] = {"name": "Webcam", "price": 2500}
```

```python
def view_products():
    if len(products) == 0:
        print("No products available")
    else:
        for pid in products:
            print(f"ID: {pid}, Name: {products[pid]['name']}, Price: {products[pid]['price']}")
```

```python
def add_product():
    pid = int(input("Enter Product ID: "))
    name = input("Enter Product Name: ")
    price = float(input("Enter Price: "))

    if pid in products:
        print("Product ID already exists!")
    else:
        products[pid] = {"name": name, "price": price}
        print("Product added successfully")
```

```python
def search_product():
    pid = int(input("Enter Product ID to search: "))
    if pid in products:
        print(products[pid])
    else:
        print("Product not found")
```

```python
load_dummy_data()
```

```python
view_products()
```

```python
def update_product():
    pid = int(input("Enter Product ID to update: "))
    if pid in products:
        name = input("Enter New Name: ")
        price = float(input("Enter New Price: "))
        products[pid]["name"] = name
        products[pid]["price"] = price
        print("Product updated")
    else:
        print("Product not found")
```

```python
def delete_product():
    pid = int(input("Enter Product ID to delete: "))
    if pid in products:
        del products[pid]
        print("Product deleted")
    else:
        print("Product not found")
```

```python
import os
from getpass import getpass
from groq import Groq

MODEL = "openai/gpt-oss-20b"  # Model ID
```

```python
os.environ["GROQ_API_KEY"] = getpass("Enter Groq API key: ").strip()
```

```python
def ai_product_details():
    if not products:
        print("No products available")
        return

    query = input("Ask anything about products: ")

    client = Groq(api_key=os.getenv("GROQ_API_KEY"))

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a helpful product assistant. "
                    "Answer using the supplied inventory and statistics.\n"
                    f"Product inventory: {products}\n"
                )
            },
            {"role": "user", "content": query}
        ]
    )

    print("\nAI Response:")
    print(response.choices[0].message.content)
```

```python
def dashboard():
    while True:
        print("\n--- PRODUCT MANAGEMENT ---")
        print("1. View Products")
        print("2. Add Product")
        print("3. Search Product")
        print("4. Update Product")
        print("5. Delete Product")
        print("6. AI Product Help (Groq)")
        print("7. Exit")
        choice = int(input("Enter choice: "))
        match choice:
            case 1:
                view_products()
            case 2:
                add_product()
            case 3:
                search_product()
            case 4:
                update_product()
            case 5:
                delete_product()
            case 6:
                ai_product_details()
            case 7:
                print("Exiting...")
                break
            case _:
                print("Invalid choice")


def main():
    load_dummy_data()
    dashboard()
```

```python
main()
```
